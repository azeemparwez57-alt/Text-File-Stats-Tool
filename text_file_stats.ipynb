```python
# read_file(filename)
def read_file(filename):
    with open(filename) as f:
        text = f.read()
    return text

#count_lines(text)
def count_lines(text):
    if text.strip() == "":
        return 0
        
    lines = text.split("\n")
    return len(lines)

#count_words(text)
def count_words(text):
    words = text.split()
    return(len(words))

#count_characters(text)
def count_characters(text):
    text = text.replace(" ", "").replace("\n", "").replace("\t", "")
    return len(text)

#longest_line(text)
def longest_line(text):
    if text.strip() == "":
        return ""
        
    long_line = ""
    text = text.split("\n")
    for i in text:
        if len(long_line) < len(i):
            long_line = i
    return long_line

#smallest_line(text)
def smallest_line(text):
    if text.strip() == "":
        return ""
        
    lines = text.split("\n")
    non_empty_lines = []
    for line in lines:
        if line.strip() != "":
            non_empty_lines.append(line)
    small_line = non_empty_lines[0]
    for i in non_empty_lines:
        if len(i) < len(small_line):
            small_line = i
    return small_line

#word_frequency(text)
def word_frequency(text):
    common_words_freq = {}
    lines = text.split("\n")
    for line in lines:
        words = line.split()
        for word in words:
            word = word.lower().strip(".,!?;:")
            if word in common_words_freq:
                common_words_freq[word] += 1
            else:
                common_words_freq[word] = 1
    return common_words_freq        

#most_common_word(text)
def most_common_word(text):
    comm_word = word_frequency(text)
    m_comm_word = ""
    large = 0
    for i in comm_word.items():
        if i[1] > large:
            large = i[1]
            m_comm_word = i[0]
    return m_comm_word

#Main program
filename = input("Enter your filename: ")
try:
    text = read_file(filename)
    print(f"Total Lines: {count_lines(text)}")
    print(f"Total Words: {count_words(text)}")
    print(f"Total Characters: {count_characters(text)}")
    long_line = longest_line(text)
    if long_line == "":
        print("Longest Line: \"No lines found.\"")
    else:
        print(f"Longest Line: {long_line}")
    small_line = smallest_line(text)
    if small_line == "":
        print("Smallest Line: \"No lines found.\"")
    else:
        print(f"Smallest Line: {small_line}")
    common_word = most_common_word(text)
    if common_word == "":
        print("Most Common Word:\" No words found.\"")
    else:
        print(f"Most Common Word: {common_word}")
except FileNotFoundError:
    print("Error: File not found. Please check the filename.")
```
